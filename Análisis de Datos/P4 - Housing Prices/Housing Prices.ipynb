```python
# Librerías a Utilizar
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Importar Data
housing_data = pd.read_csv("./Data/housing_data_CDMX.csv")
housing_data.head()
```

```python
housing_data.info()
```

```python
housing_data.describe()
```

```python
# Eliminar columnas innecesarias
housing_data.drop(columns=["lat-lon", "currency", "price", "price_aprox_local_currency", "price_per_m2"], inplace=True)
housing_data.head()
```

```python
# Valores Nulos y Duplicados
housing_data.isnull().sum()
```

```python
housing_data.drop_duplicates(inplace=True)
```

### Gráficos 

```python
# Ventas por Tipo de Inmueble
plt.figure(figsize=(4,4))

sns.countplot(data=housing_data, x='property_type', order=housing_data['property_type'].value_counts().index)
plt.title('Número de Ventas por Tipo de Inmueble')
plt.xlabel('Tipo de Inmueble')
plt.ylabel('Número de Ventas')
plt.xticks(rotation=45)
plt.show()
```

```python
# Distrución de Precios por Tipo de Inmueble
plt.figure(figsize=(8,6))
sns.boxplot(data=housing_data, x="property_type", y="price_aprox_usd")

plt.title("Distribución de Precios por Tipo de Inmueble")
plt.xlabel("Tipo de Inmueble")
plt.ylabel("Precio por m² (USD)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Distribución de Precios
plt.figure(figsize=(8,6))

sns.histplot(housing_data['price_aprox_usd'], bins=30, kde=True)
plt.title('Distribución de Precios de Viviendas')

plt.xlabel('Precio Aproximado (USD)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# Conversión a logaritmo
def log_transform(price):
    return np.log(price + 1)

prices = ["price_aprox_usd", "surface_total_in_m2", "surface_covered_in_m2", "price_usd_per_m2"]

for col in prices:
    housing_data[f'log_{col}'] = housing_data[col].apply(log_transform)
housing_data.head()
```

```python
# Distribución de Precios
plt.figure(figsize=(8,6))
sns.histplot(housing_data["log_price_aprox_usd"], bins=30, kde=True)

plt.title("Distribución de Precios (Logaritmo)")
plt.xlabel("Logaritmo del Precio Aproximado (USD)")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# Diagrama de dispersión de Precio vs. Superficie
plt.figure(figsize=(8,4))
sns.scatterplot(data=housing_data, x="log_surface_total_in_m2", y="log_price_aprox_usd", alpha=0.5)
plt.title("Precio vs. Superficie Total")
plt.xlabel("Superficie Total (m²)")
plt.ylabel("Precio Aproximado (USD)")
plt.show()
```

```python
# Heatmap de Correlación
plt.figure(figsize=(8,4))

corr = housing_data[['log_price_aprox_usd', 'log_surface_total_in_m2', 'log_surface_covered_in_m2', 'log_price_usd_per_m2']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Matriz de Correlación')
plt.show()
```

```python
# Graficar Distribuciones
def plot_distribution(data, column, title, xlabel):
    plt.figure(figsize=(4,4))
    sns.histplot(data[column], bins=30, kde=True)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel('Frecuencia')
    plt.show()

col = ["log_price_aprox_usd", "log_surface_total_in_m2", "log_surface_covered_in_m2", "log_price_usd_per_m2"]

for c in col:
    plot_distribution(housing_data, c, f'Distribución de {c}', c)
```
