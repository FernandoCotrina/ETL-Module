```python
# Import Necesary Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
data = pd.read_csv("./Data/student_data.csv")
```

Analysis of Student Performance Data

```python
data.head()
```

```python
data.info()
```

Treatment of Categorical Columns

```python
categorical_cols = ["gender", "school_type", "parent_education", "internet_access", "extra_activities", "study_method"]

for col in categorical_cols:
    data[col] = data[col].astype('category')

data.info()
```

#### Data Visualization

```python
# Violin Plot of Final Grades by Study Method
plt.figure(figsize=(7,3))

sns.violinplot(data=data, x="study_method", y="overall_score", hue="study_method", palette="viridis")
plt.title("Distribution of Final Grades by Study Method")
plt.xlabel("Study Method")
plt.ylabel("Final Grade")

plt.show()
```

```python
# Boxplot of Notes of Students by Gender and Course
def boxplot_notes(col):
    plt.figure(figsize=(8,4))

    sns.boxplot(data=data, x="gender", y=col, hue="gender", palette="Set2")
    plt.title(f"Boxplot of {col.capitalize()} by Gender")

    plt.show()

courses = ["math_score", "science_score", "english_score", "overall_score"]
for col in courses:
    boxplot_notes(col)
```

```python
# Mean Overall Score by Private vs Public School
plt.figure(figsize=(6,4))

sns.barplot(data=data, x="school_type", y="overall_score", hue="school_type", palette="pastel")
plt.title("Mean Overall Score by School Type")

plt.show()
```
